```python
# Databricks Notebook

# ==========================================
# Configuration
# ==========================================

catalog_name = "my_catalog"

schemas = [
    "landing",
    "bronze",
    "silver",
    "gold"
]

catalog_comment = "Enterprise data platform catalog"

schema_comments = {
    "landing": "Raw files received from source systems",
    "bronze": "Raw ingested data with minimal transformations",
    "silver": "Cleansed and conformed business data",
    "gold": "Aggregated and curated business datasets"
}

# ==========================================
# Create Catalog
# ==========================================

spark.sql(f"""
CREATE CATALOG IF NOT EXISTS {catalog_name}
COMMENT '{catalog_comment}'
""")

print(f"Catalog created or already exists: {catalog_name}")

# ==========================================
# Create Schemas
# ==========================================

for schema in schemas:
    spark.sql(f"""
    CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema}
    COMMENT '{schema_comments.get(schema, "")}'
    """)

    print(f"Schema created or already exists: {catalog_name}.{schema}")

# ==========================================
# Verify
# ==========================================

display(
    spark.sql(f"SHOW SCHEMAS IN {catalog_name}")
)
```
